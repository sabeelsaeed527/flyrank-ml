```python
import duckdb
import pandas as pd
from google.colab import userdata

# Get HF token from Colab Secret
HF_TOKEN = userdata.get("HF_TOKEN")

# Check token
if not HF_TOKEN:
    raise ValueError("HF_TOKEN not found. Check Colab Secrets.")

# Create DuckDB connection
con = duckdb.connect()

# Authenticate DuckDB with Hugging Face
con.execute(f"""
CREATE SECRET hf_secret (
    TYPE HUGGINGFACE,
    TOKEN '{HF_TOKEN}'
)
""")

print("✅ DuckDB connected to Hugging Face!")
```

```python
# Show all available tables
print(con.execute("SHOW ALL TABLES").fetchdf())
```

```python
# List the files available in the FlyRank warehouse dataset
from huggingface_hub import HfApi

api = HfApi(token=HF_TOKEN)

files = api.list_repo_files(
    repo_id="FlyRank/internship-warehouse",
    repo_type="dataset"
)

for f in files:
    print(f)
```
