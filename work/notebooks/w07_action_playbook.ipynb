## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

outputs_dir = Path("../outputs")
figures_dir = Path("../figures")
outputs_dir.mkdir(parents=True, exist_ok=True)
figures_dir.mkdir(parents=True, exist_ok=True)

# Simulated or loaded data frame structure for the action queue
data = {
    'page_id': ['p_01', 'p_02', 'p_03', 'p_04'],
    'predicted_decay': [0.85, 0.78, 0.62, 0.20],
    'traffic_tier': ['High', 'Low', 'High', 'Low'],
    'engagement_score': [0.25, 0.40, 0.15, 0.85]
}
df_queue = pd.DataFrame(data)

conditions = [
    (df_queue['predicted_decay'] > 0.7) & (df_queue['traffic_tier'] == 'High'),
    (df_queue['predicted_decay'] > 0.7) & (df_queue['traffic_tier'] == 'Low'),
    (df_queue['engagement_score'] < 0.3)
]
actions = [
    'Refresh metadata & code snippets',
    'Archive or redirect URL',
    'Content structural layout update'
]
reasons = [
    'HIGH_TRAFFIC_STALE',
    'DECAY_RISK_LOW_TRAFFIC',
    'LOW_ENGAGEMENT_MISMATCH'
]

df_queue['recommended_action'] = np.select(conditions, actions, default='Monitor performance')
df_queue['reason_code'] = np.select(conditions, reasons, default='STABLE_PASS')
df_queue = df_queue.sort_values(by='predicted_decay', ascending=False)
print(f"Action queue generated with {len(df_queue)} rows.")
```
